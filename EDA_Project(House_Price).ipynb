```python
df_cleaned = df_cleaned[['date', 'price', 'bedrooms', 'bathrooms', 'sqft_living', 'sqft_lot',
       'floors', 'waterfront', 'view', 'condition', 'sqft_above',
       'sqft_basement', 'yr_built', 'yr_renovated', 'street', 'city',
       'statezip']]
```

```python
df_cleaned
```

```python
df_cleaned.value_counts('statezip')
```

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()

df_cleaned['city'] = le.fit_transform(df_cleaned['city'])
df_cleaned['street'] = le.fit_transform(df_cleaned['street'])
df_cleaned['statezip'] = le.fit_transform(df_cleaned['statezip'])
```

```python
df_cleaned
```

```python
# Correlation of all numerical columns with price
correlation = df_cleaned.corr(numeric_only=True)['price'].sort_values(ascending=False)

print(correlation)
```

```python
df_final = df_cleaned[['date', 'price', 'bedrooms', 'bathrooms', 'sqft_living', 'sqft_lot',
       'floors', 'waterfront', 'view', 'sqft_above',
       'sqft_basement']]
```

```python
df_final
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

numeric_features = [
   'price', 'bedrooms', 'bathrooms', 'sqft_living', 'sqft_lot',
       'floors', 'waterfront', 'view', 'sqft_above',
       'sqft_basement'
]

df_final[numeric_features] = scaler.fit_transform(df_cleaned[numeric_features])
```
